# Bài 1

````python
from transformers import pipeline
````

````python
mask_filler = pipeline('fill-mask')
````

````python
input_sentence = "Hanoi is the <mask> of Vietnam."
````

````python
predictions = mask_filler(input_sentence, top_k=5)
````

````python
print(f"Câu gốc: {input_sentence}")
for pred in predictions:
    print(f"Dự đoán: '{pred['token_str']}' với độ tin cậy: {pred['score']:.4f}")
    print(f" -> Câu hoàn chỉnh: {pred['sequence']}")
````

**Câu hỏi 1: Mô hình đã dự đoán đúng từ "capital" không?**

Có, mô hình đã dự đoán đúng từ "capital" với độ tin cậy rất cao (93.41%). Kết quả cho thấy:
- Top 1: "capital" (93.41%) - **Dự đoán chính xác**
- Top 2: "Republic" (3.00%)
- Top 3: "Capital" (1.05%) - Viết hoa, cùng nghĩa nhưng ít phổ biến hơn

Mô hình BERT hiểu rõ quan hệ ngữ nghĩa "Hanoi - capital - Vietnam" từ dữ liệu pre-training.

---

**Câu hỏi 2: Tại sao các mô hình Encoder-only như BERT lại phù hợp cho tác vụ này?**

Mô hình Encoder-only (BERT) phù hợp cho tác vụ fill-mask vì:

**1. Bidirectional Context (Ngữ cảnh hai chiều)**
- BERT đọc cả từ trái sang phải VÀ phải sang trái đồng thời
- Với câu "Hanoi is the `<mask>` of Vietnam", BERT thấy:
  - Context trái: "Hanoi is the"
  - Context phải: "of Vietnam"
- Kết hợp hai chiều giúp dự đoán chính xác hơn

**2. Masked Language Modeling (MLM) Pre-training**
- BERT được huấn luyện sẵn với tác vụ dự đoán từ bị che (exactly fill-mask)
- Trong pre-training, 15% tokens bị mask ngẫu nhiên
- Mô hình học cách dự đoán từ dựa vào ngữ cảnh xung quanh
- Fill-mask là tác vụ native của BERT

**3. Deep Contextual Understanding**
- 12 layers Transformer encoders (BERT-base) tạo ra biểu diễn ngữ cảnh sâu
- Mỗi layer học các đặc trưng khác nhau:
  - Layer thấp: Syntax, POS tags
  - Layer giữa: Semantic relationships
  - Layer cao: Task-specific patterns
- Giúp hiểu quan hệ "capital city - country"


**Kết luận:** BERT được thiết kế riêng cho fill-mask với MLM objective, bidirectional context, và hiệu quả cao - đây là tác vụ mà BERT excel nhất!

**Câu hỏi 2: Tại sao các mô hình Decoder-only như GPT lại phù hợp cho tác vụ này?**

Mô hình Decoder-only (GPT) phù hợp cho text generation vì:

**1. Autoregressive Generation (Sinh tuần tự)**
- GPT sinh text từ trái sang phải, từng token một
- Mỗi token mới dựa vào tất cả tokens trước đó
- Quy trình: Prompt → token₁ → token₂ → ... → token_n
- Ví dụ: "The best thing" → "about" → "learning" → "NLP" → ...
- **Phù hợp với bản chất của text generation**: Con người cũng viết/nói tuần tự từ trái sang phải

**2. Causal Attention (Chỉ nhìn về quá khứ)**
- GPT chỉ được phép nhìn các tokens ở phía trước (left context)
- Attention mask hình tam giác ngăn model "nhìn trước"
- Ngăn chặn "cheating" - model không biết trước văn bản sẽ sinh ra
- **Đảm bảo tính khả thi**: Khi sinh token thứ 10, chưa có token thứ 11

**3. Next Token Prediction Pre-training**
- GPT được train với objective: dự đoán token tiếp theo
- Training data: Triệu tỷ câu từ Internet
- Model học patterns: grammar, semantics, world knowledge
- **Text generation là tác vụ native của GPT** (giống BERT với fill-mask)

**4. Long Context Handling**
- Decoder architecture duy trì context dài qua nhiều layers
- Mỗi layer tinh chỉnh representation dựa trên context trước đó
- GPT-2 (1.5B params) có khả năng nhớ context ~1024 tokens
- **Sinh văn bản dài và mạch lạc**


**Kết luận:** GPT được thiết kế riêng cho text generation với causal attention, autoregressive generation, và next token prediction objective. Đây là kiến trúc tối ưu cho tác vụ sinh văn bản tự do

````python
import torch
from transformers import AutoTokenizer, AutoModel
````

````python
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
````

````python
sentence = "This is the sample sentence."
````

````python
input = tokenizer(sentence, padding=True, truncation=True, return_tensors="pt")
````

````python
with torch.no_grad():
    outputs = model(**input)    
````

````python
last_hidden_states = outputs.last_hidden_state
````

````python
attention_mask = input['attention_mask']
mask_expanded = attention_mask.unsqueeze(-1).expand(last_hidden_states.size()).float()
sum_embeddings = torch.sum(last_hidden_states * mask_expanded, 1)
sum_mask = torch.clamp(mask_expanded.sum(1), min=1e-9)
sentence_embeddings = sum_embeddings / sum_mask
````

````python
print("Vector biểu diễn của câu:")
print(sentence_embeddings)
print("\nKích thước của vector:",   sentence_embeddings.shape)
````

**Câu hỏi 1: Kích thước (chiều) của vector biểu diễn là bao nhiêu? Con số này tương ứng với tham số nào của mô hình BERT?**

Kích thước của vector biểu diễn là **768 chiều** (shape: `torch.Size([1, 768])`).

Con số này tương ứng với tham số **`hidden_size`** của mô hình BERT:
- `model.config.hidden_size = 768`
- Đây là số chiều của hidden states trong mỗi layer của BERT-base

**Chi tiết kiến trúc BERT-base:**
```
- Vocabulary size: 30,522 tokens
- Hidden size: 768 dimensions (vector embedding cho mỗi token)
- Number of layers: 12 Transformer encoder layers
- Number of attention heads: 12 heads (mỗi head 64 dimensions)
- Intermediate size: 3,072 (FFN hidden layer)
- Max sequence length: 512 tokens
```

**Giải thích shape:**
- `last_hidden_states`: `[1, 8, 768]`
  - Batch size = 1 (1 câu)
  - Sequence length = 8 tokens ("[CLS] this is the sample sentence . [SEP]")
  - Hidden size = 768 (mỗi token có vector 768 chiều)
  
- `sentence_embeddings`: `[1, 768]`
  - Batch size = 1
  - Hidden size = 768 (vector đại diện cho cả câu sau khi mean pooling)

**Tại sao 768?**
- Con số 768 được chọn để cân bằng giữa capacity (khả năng biểu diễn) và efficiency (tốc độ tính toán)
- 768 = 12 heads × 64 dimensions/head (dễ chia cho multi-head attention)
- BERT-large sử dụng 1024 dimensions cho hidden size

**Câu hỏi 2: Tại sao chúng ta cần sử dụng attention_mask khi thực hiện Mean Pooling?**

Chúng ta cần `attention_mask` khi thực hiện Mean Pooling để **loại bỏ padding tokens khỏi phép tính trung bình**.

**Lý do chi tiết:**

**1. Vấn đề với Padding Tokens**

Khi xử lý batches với câu có độ dài khác nhau, tokenizer tự động thêm padding:
```python
# Ví dụ:
Sentence 1: "This is short"          → 5 tokens + 3 padding = 8 tokens
Sentence 2: "This is a longer one"   → 7 tokens + 1 padding = 8 tokens
```

Padding tokens có giá trị `[PAD]` (token_id = 0) và không mang ý nghĩa ngữ nghĩa. Nếu tính mean bao gồm padding:
```python
# SAI - Bao gồm padding:
mean = sum(all 8 tokens) / 8
# → Giá trị embedding bị pha loãng bởi padding

# ĐÚNG - Loại bỏ padding:
mean = sum(5 real tokens) / 5
# → Chỉ tính trung bình các tokens thực sự
```

**2. Vai trò của attention_mask**

`attention_mask` là tensor binary (0 hoặc 1):
```python
attention_mask = [1, 1, 1, 1, 1, 0, 0, 0]
#                 ↑ Real tokens  ↑ Padding
```

- `1`: Token thực (cần tính vào mean)
- `0`: Padding token (bỏ qua)


**Kết luận:** `attention_mask` đảm bảo Mean Pooling chỉ tính trên tokens thực sự, loại bỏ ảnh hưởng của padding, giúp sentence embedding chính xác và có ý nghĩa ngữ nghĩa!
